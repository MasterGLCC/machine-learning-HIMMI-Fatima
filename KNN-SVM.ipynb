```python
import math
```

```python

# Datasset
# Ici j'ai créé un petitt dataset qui représente des étudiants / projets PFE
# Il contient 3 caractéristiques numériques + une classe 
# Les 3 premières valeurs = niveau dans chaque domaine
# La dernière valeur = catégorie du projet PFE (Web, ML ou UI/UX)
data = [
    [8, 2, 3, "Application Web"],
    [7, 1, 2, "Application Web"],
    
    [3, 9, 2, "Machine Learning"],
    [2, 8, 3, "Machine Learning"],
    
    [2, 3, 9, "UI/UX Design"],
    [1, 2, 8, "UI/UX Design"],
    
    [6, 5, 4, "Application Web"],
    [4, 6, 3, "Machine Learning"]
]

# Fonction de distance euclidiennne
# Ici je calcule la distance entre deux points pour voir à quel point ils se ressemblent.
def distance(p1, p2):
    return math.sqrt(
        (p1[0] - p2[0])**2 +   
        (p1[1] - p2[1])**2 +  
        (p1[2] - p2[2])**2    
    )

#  ALGORITHME KNN 
# prédire la classe d'un nouveau point en regardant ses k voisins les plus proches
def knn(data, new_point, k=3):
    
    # Je vais stocker ici les distances avec leur classe
    distances = []
    #  CALCUL DES DISTANCES
    # Je compare le nouveau point avec chaque élément du dataset

    for row in data:
        # row[:3] = les 3 features 
        d = distance(new_point, row[:3])
        
        # je stocke la distance + la classe associée
        distances.append((d, row[3]))
        
    # Je trie du plus proche au plus loin
    distances.sort(key=lambda x: x[0])
    
    # Je prends les k voisins les plus proches
    neighbors = distances[:k]

    # vote majoritaire
    votes = {}

    for _, label in neighbors:
        if label in votes:
            votes[label] += 1
        else:
            votes[label] = 1
            
    # Je retourne la classe qui a le plus de votes
    return max(votes, key=votes.get)

# C’est le nouvel étudiant dont on veut prédire la classe

Fatima = [7, 6, 3]
# Prédiction
resultat = knn(data, Fatima, k=3)

#  Affichagee
print(" Projet recommandé :", resultat)
```

```python
from sklearn import svm

#  Dataset
data = [
    [8, 2, 3, "Application Web"],
    [7, 1, 2, "Application Web"],
    
    [3, 9, 2, "Machine Learning"],
    [2, 8, 3, "Machine Learning"],
    
    [2, 3, 9, "UI/UX Design"],
    [1, 2, 8, "UI/UX Design"],
    
    [6, 5, 4, "Application Web"],
    [4, 6, 3, "Machine Learning"]
]

# Séparation des features et des labels

# X = les caractéristiques 
# y = la catégorie du projet (la réponse qu'on veut prédire)

X = [row[:3] for row in data]
y = [row[3] for row in data]

# J'utilise SVM avec un kernel linéaire

model = svm.SVC(kernel='linear')

#  Entraînement du modèle
# Ici le modèle apprend à partir des données pour comprendre la relation entre les compétences et le type de projet

model.fit(X, y)

# Nouveau cas 
Fatima = [[7, 6, 3]]
# Prédiction
# Le modèle va prédire le type de projet le plus adapté

resultat = model.predict(Fatima)

print(" Projet recommandé :", resultat[0])
```
