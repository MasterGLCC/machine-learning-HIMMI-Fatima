```python
#ALGO CART


# Chaquee ligne représente un étudiant
# Les deux premiers nmbres représentent le niveau en Web et ML
# La dernière valeur représente le type de projet recommandé

data = [
    [8, 2, "Web"],
    [7, 1, "Web"],

    [2, 8, "ML"],
    [3, 9, "ML"]
]


# J’ai créé cette fct pour calculerr le Gini qui permet de mesurer si les données sont mélangées ou non
def gini(group):

    #  je récupère le nombre total d’éléments
    total = len(group)

    # J’utilise un dictionnaire pour compter les classes
    classes = {}

    # Je parcours chaque ligne du groupe
    for row in group:

        # récupérer la classe
        label = row[2]

        #  je compte combien de fois chaque classe apparaît
        classes[label] = classes.get(label, 0) + 1

    # Variable pour stocker le score Gini
    score = 0

    # J’applique la formule du Gini
    for count in classes.values():

        # Je calcule la probabilité
        p = count / total

        # J’ajoute p² au score
        score += p * p

    # Je retourne le score final
    return 1 - score


# J’ai créé cette fonction pour séparer les données

def split(data, feature, value):

    left = []
    right = []

    # Je parcours toutes les lignes du dataset
    for row in data:

        # Si la condition est vraie
        if row[feature] <= value:

            # J’ajoute la ligne dans le groupe left
            left.append(row)

        else:

            # Sinon je l’ajoute dans le groupe right
            right.append(row)

    # Je retourne les deux groupes
    return left, right


# algo CART
def cart(data, new_point):

    # Variables pour garder le meilleur split trouvé
    best_gini = 1
    best_feature = 0
    best_value = 0

    # je teste chaque feature
    for feature in range(2):

        #je teste plusieurs valeurs possibles
        for row in data:

            # Je récupère la valeur actuelle
            value = row[feature]

            # Je sépare les données
            left, right = split(data, feature, value)

            # Je calcule le score Gini moyen
            score = (gini(left) + gini(right)) / 2

            # Si le score est meilleur
            if score < best_gini:

                # Je garde ce split
                best_gini = score
                best_feature = feature
                best_value = value

    # apres je refais le meilleur split trouvé
    left, right = split(data, best_feature, best_value)

    # Je vérifie dans quel groupe va le nouvel étudiant
    if new_point[best_feature] <= best_value:

        group = left

    else:

        group = right

    # j’utilise le vote majoritaire
    votes = {}

    # Je compte les classes du groupe choisi
    for row in group:

        label = row[2]

        votes[label] = votes.get(label, 0) + 1

    # Je retourne la classe qui a le plus de votes
    return max(votes, key=votes.get)


#  j’ai créé un nouvel étudiant pour le test
Fatima = [7, 3]

# Affichage 
print("Projet recommandé :", cart(data, Fatima))
```

```python
import math

data = [
    [8, 2, "Web"],
    [7, 1, "Web"],

    [2, 8, "ML"],
    [3, 9, "ML"]
]


# J’ai créé cette fct pour calculer l’entropie, plus l’entropie est petite, plus le groupe est pur

def entropy(group):

    # Je récupère le nombre total d’éléments
    total = len(group)

    # Dictionnaire pour compter les classes
    classes = {}

    # Je parcours chaque ligne
    for row in group:

        # Je récupère la classe
        label = row[2]

        # Je compte combien de fois chaque classe apparaît
        classes[label] = classes.get(label, 0) + 1

    # Variable de l’entropie
    ent = 0

    # J’applique la formule de l’entropie
    for count in classes.values():

        # Calcul de la probabilité
        p = count / total

        # Formule : -p log2(p)
        ent -= p * math.log2(p)

    # Je retourne le résultat final
    return ent


# J’ai créé cette fonction pour séparer les données, chaque valeur de la feature aura son propre groupe

def split(data, feature):

    groups = {}

    # Je parcours les lignes du dataset
    for row in data:

        # Je récupère la valeur de la feature
        key = row[feature]

        # Si la valeur n’existe pas encore
        if key not in groups:
            groups[key] = []

        # J’ajoute la ligne dans son groupe
        groups[key].append(row)

    # Je retourne les groupes
    return groups


# Fonction principale de l’algorithme ID3
def id3(data, new_point):

    # Variables pour garder la meilleure feature
    best_feature = 0
    best_entropy = 999

    # Je teste chaque feature
    for feature in range(2):

        # Je sépare les données
        groups = split(data, feature)

        # Variable pour stocker l’entropie totale
        score = 0

        # Je calcule l’entropie moyenne
        for group in groups.values():

            score += (len(group) / len(data)) * entropy(group)

        # Si le score est meilleur
        if score < best_entropy:

            # Je garde cette feature
            best_entropy = score
            best_feature = feature

    # Je récupère toutes les valeurs possibles
    values = [row[best_feature] for row in data]

    # Je cherche la valeur la plus proche du nouvel étudiant
    closest = min(values, key=lambda x: abs(x - new_point[best_feature]))

    # Je prends le groupe correspondant
    group = [row for row in data if row[best_feature] == closest]

    # Vote majoritaire
    votes = {}

    # Je compte les classes du groupe
    for row in group:

        label = row[2]

        votes[label] = votes.get(label, 0) + 1

    # Je retourne la classe majoritaire
    return max(votes, key=votes.get)


# un nouvel étudiant 
Fatima = [7, 3]

# Affichage 
print("Projet recommandé :", id3(data, Fatima))
```

```python
import math

# ALGO C4.5

data = [
    [8, 2, "Web"],
    [7, 1, "Web"],

    [2, 8, "ML"],
    [3, 9, "ML"]
]


# Fonction Entropie
# Si tous les éléments sont de la même classe → entropie faible
# Si les classes sont mélangées → entropie élevée

def entropy(group):

    # Je calcule le nombre total d’éléments dans le groupe
    total = len(group)

    # J’utilise un dictionnaire pour compter les classes
    classes = {}

    # Je parcours chaque élément du groupe
    for row in group:

        # Je récupère la classe 
        label = row[2]

        # Je compte combien de fois chaque classe apparaît
        classes[label] = classes.get(label, 0) + 1

    # Variable pour stocker l’entropie
    ent = 0

    # J’applique la formule de l’entropie
    for count in classes.values():

        # Je calcule la probabilité de chaque classe
        p = count / total

        # J’applique la formule : - p * log2(p)
        ent -= p * math.log2(p)

    # Je retourne l’entropie totale du groupe
    return ent


# Fct pour séparer les données

def split(data, feature, value):
    left = []
    right = []

    # Je parcours toutes les lignes 
    for row in data:

        # Si la valeur est petite ou égale au seuil
        if row[feature] <= value:

            # Je l’ajoute au groupe gauche
            left.append(row)

        else:

            # Sinon je l’ajoute au groupe droit
            right.append(row)

    # Je retourne les deux groupes
    return left, right


# Algorithme C4.5 (version simplifiée)
# Objectif : trouver la meilleure séparation des données

def c45(data, new_point):

    # Je stocke la meilleure entropie trouvée
    best_entropy = 999

    # Je stocke la meilleure feature
    best_feature = 0

    # Je stocke la meilleure valeur de séparation
    best_value = 0

    # Je teste chaque feature 
    for feature in range(2):

        # Je teste plusieurs valeurs possibles
        for row in data:

            # Je récupère la valeur actuelle
            value = row[feature]

            # Je sépare les données selon cette règle
            left, right = split(data, feature, value)

            # Je calcule l’entropie moyenne des deux groupes
            score = (entropy(left) + entropy(right)) / 2

            # Je garde la meilleure séparation (entropie la plus petite)
            if score < best_entropy:

                best_entropy = score
                best_feature = feature
                best_value = value

    # Une fois la meilleure séparation trouvée
    left, right = split(data, best_feature, best_value)

    # Je décide dans quel groupe va le nouvel étudiant
    if new_point[best_feature] <= best_value:

        group = left

    else:

        group = right

    # Je fais un vote majoritaire dans le groupe choisi
    votes = {}

    for row in group:

        label = row[2]

        # Je compte les classes
        votes[label] = votes.get(label, 0) + 1

    # Je retourne la classe la plus fréquente
    return max(votes, key=votes.get)


# Nouvel étudiant
Fatima = [7, 3]

# Résultat 
print("Projet recommandé :", c45(data, Fatima))
```
