```python
# DBSCAN


import math

# Chaque ligne représente un étudiant
# Le premier nombre représente le niveau en Web
# Le deuxième nombre représente le niveau en ML

data = [
    [8, 2],
    [7, 1],

    [2, 8],
    [3, 9]
]


# Fonction pour calculer la distance entre deux points (distance euclidienne)

def distance(p1, p2):

    return math.sqrt(

        # Différence sur la première caractéristique (Web)
        (p1[0] - p2[0])**2 +

        # Différence sur la deuxième caractéristique (ML)
        (p1[1] - p2[1])**2
    )


# Fonction DBSCAN pour créer des groupes de points proches

def dbscan(data, eps):

    # Liste qui va contenir tous les clusters trouvés
    clusters = []

    # Je parcours chaque point 
    for point in data:

        # Je crée un nouveau cluster vide
        cluster = []

        # Je compare ce point avec tous les autres points
        for other in data:

            # Je calcule la distance entre les deux points
            d = distance(point, other)

            # Si la distance est inférieure ou égale à eps donc les deux points sont considérés proches
            if d <= eps:

                # J’ajoute le point dans le même cluster
                cluster.append(other)

        # j’ajoute le cluster dans la liste finale
        clusters.append(cluster)

    return clusters


# eps représente la distance maximale entre deux points
# Si deux points dépassent cette distance alors ils ne seront pas dans le même groupe

print(dbscan(data, 2))
```
