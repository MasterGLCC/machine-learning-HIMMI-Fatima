```python
# Q-Learning


import random

# Domaines possibles
states = ["Début"]

# Actions possibles
actions = ["Web", "IA", "Réseau"]

# Table Q
# Cette table va mémoriser les scores des choix
Q = {
    "Début": {
        "Web": 0,
        "IA": 0,
        "Réseau": 0
    }
}

# Récompenses, le meilleur choix est Web
rewards = {
    "Web": 10,
    "IA": 5,
    "Réseau": 2
}


# Fonction Q-Learning 
def q_learning():

    # Je commence au début
    state = "Début"

    # Je fais plusieurs essais
    for i in range(10):

        # Choisir une action aléatoire
        action = random.choice(actions)

        # Récupérer la récompense du choix
        reward = rewards[action]

        # Mettre à jour le score dans la table Q
        Q[state][action] = reward


# Lancer l’apprentissage
q_learning()

# Affichage 
print(Q)
```
