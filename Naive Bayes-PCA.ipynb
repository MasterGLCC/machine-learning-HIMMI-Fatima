```python
# Naive Bayes


# Chaque ligne représente un étudiant avec ses compétences
# Les deux premiers nombres représentent le niveau en Web et en ML
# La dernière valeur représente la classe du projet

data = [
    [8, 2, "Web"],
    [7, 1, "Web"],

    [2, 8, "ML"],
    [3, 9, "ML"]
]


# Fct Naive Bayes 
# L’objectif est de prédire la classe d’un nouvel étudiant, en utilisant les probabilités des classes 

def naive_bayes(data, new_point):

    # Je crée un dictionnaire pour compter les classes
    counts = {}

    # Je calcule le nombre total d’exemples
    total = len(data)

    # Je parcours tout le dataset pour compter les classes
    for row in data:

        # Je récupère la classe (Web ou ML)
        label = row[2]

        # Je compte combien de fois chaque classe apparaît
        counts[label] = counts.get(label, 0) + 1


    # Je crée un dictionnaire pour stocker les probabilités finales
    probs = {}


    # Je vais maintenant calculer un score pour chaque classe
    for label in counts:

        # Je calcule la probabilité initiale de la classe 
        prior = counts[label] / total

        # Je commence le score 
        score = prior


        # Mtn je cmpare les données de la classe avec le nouvel étudiant

        for row in data:

            # Je vérifie si la ligne appartient à la classe actuelle
            if row[2] == label:

                # Si la première caractéristique (Web) est identique
                if row[0] == new_point[0]:
                    # J’augmente légèrement le score
                    score += 0.1

                # Si la deuxième caractéristique (ML) est identique
                if row[1] == new_point[1]:
                    # J’augmente encore le score
                    score += 0.1


        # Je stocke le score final de cette classe
        probs[label] = score


    # Je retourne la classe qui a le score le plus élevé
    return max(probs, key=probs.get)


# Nouveau étudiant
Fatima = [7, 3]

# J’affiche le résultat
print("Projet recommandé :", naive_bayes(data, Fatima))
```

```python
#PCA

import math
# Chaque ligne représente un étudiant
# Chaque étudiant a 3 features 

data = [
    [8, 2, 3],
    [7, 1, 2],
    [2, 8, 9],
    [3, 9, 8]
]


# calcul de la moyenne (la moyenne de chaaque featuure)

def mean(data):

    result = []

    # Je parcours chaque colonne du dataset
    # len(data[0]) = nombre de features (3)
    for i in range(len(data[0])):

        total = 0

        # Je parcours toutes les lignes pour calculer la somme de la colonne i
        for row in data:
            total += row[i]

        # Je calcule la moyenne de cette colonne
        result.append(total / len(data))

    # Je retourne une liste contenant les moyennes de chaque colonne
    return result

# centrage des données
# Cette étape consiste à soustraire la moyenne de chaque valeur

def center(data, means):

    centered = []

    # Je parcours chaque ligne
    for row in data:

        new_row = []

        # Je parcours chaque feature de l’étudiant
        for i in range(len(row)):

            # Je fais  valeur - moyenne pour centrerr les donnees
            new_row.append(row[i] - means[i])

        # J’ajoute la nouvelle ligne centrée dans la liste finale
        centered.append(new_row)

    return centered


# PCA pour réduire 3 dimensions en 1 seule dimension (PC1)

def pca(data):

    # Je commence par calculer les moyennes des colonnes
    means = mean(data)

    # Ensuite je centre les données (soustraction des moyennes)
    centered = center(data, means)

    # Je crée une liste pour stocker les résultats finaux
    result = []

    # Je parcours chaque ligne des données centrées
    for row in centered:

        # je transforme les 3 features en une seule valeur (PC1)
        pc1 = row[0] + row[1] + row[2]

        # J’ajoute cette nouvelle représentation dans la liste finale
        result.append([pc1])

    # Je retourne les données réduites 
    return result

# Test
print("Données réduites :", pca(data))
```
